## Exploratory Data Analysis & Statistical Hypothesis Testing

**Objective:** Uncover key distributional properties, evaluate statistical hypotheses, and identify collinearity in the benchmark telemetry dataset.

```python
import os

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from scipy import stats

# Plotting style configuration
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['figure.dpi'] = 120
print('Visualization libraries initialized.')
```

### 1. Ingest Clean Dataset

```python
clean_path = '../data/clean/clean.parquet'
if not os.path.exists(clean_path):
    clean_path = '../data/raw/attenomy_telemetry_raw.csv'
df = pd.read_parquet(clean_path) if clean_path.endswith('.parquet') else pd.read_csv(clean_path)
print(f'Loaded dataset: {df.shape[0]:,} rows x {df.shape[1]} columns')
df.info()
```

### 2. Target Variable Distribution

```python
target_col = 'is_fraud_or_abuse'
if target_col in df.columns:
    print(df[target_col].value_counts(normalize=True))
    fig, ax = plt.subplots(figsize=(6, 4))
    sns.countplot(data=df, x=target_col, palette=['#10b981', '#ef4444'], ax=ax)
    ax.set_title('Target Distribution : Normal vs Fraud/Abuse', fontweight='bold')
    plt.show()
```

### 3. Statistical Hypothesis Testing
Perform formal hypothesis testing (e.g. Chi-Square test of independence for device OS vs fraud).

```python
if 'device_os' in df.columns and target_col in df.columns:
    contingency_table = pd.crosstab(df['device_os'], df[target_col])
    chi2, p_val, dof, _ = stats.chi2_contingency(contingency_table)
    print(f'Chi2 Statistic: {chi2:.4f}')
    print(f'p-value: {p_val:.4e}')
    print(f'Degrees of Freedom: {dof}')
    if p_val < 0.05:
        print('Result: Reject H0 - Significant association between Device OS and Fraud rate.')
    else:
        print('Result: Fail to reject H0 - No significant association observed.')
```
